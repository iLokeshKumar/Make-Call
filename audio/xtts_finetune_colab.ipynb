```python
# 4 — Copy dataset to Colab local storage (faster I/O)
import shutil, os
from pathlib import Path

DATASET_DIR = Path('/content/dataset')
OUT_DIR = Path('/content/xtts_finetuned')
OUT_DIR.mkdir(exist_ok=True)

if DATASET_DIR.exists():
    shutil.rmtree(DATASET_DIR)
shutil.copytree(DRIVE_DATASET, DATASET_DIR)

wavs = list((DATASET_DIR / 'wavs').glob('*.wav'))
print(f'Dataset: {len(wavs)} WAV files')
with open(DATASET_DIR / 'metadata.csv') as f:
    for i, line in enumerate(f):
        print(' ', line.strip())
        if i >= 3: break
```

```python
# 5 — Fine-tune XTTS v2 GPT decoder
import os
os.environ['COQUI_TOS_AGREED'] = '1'

from trainer import Trainer, TrainerArgs
from TTS.tts.configs.xtts_config import XttsConfig
from TTS.tts.datasets import load_tts_samples
from TTS.tts.models.xtts import Xtts, XttsAudioConfig, XttsArgs
from TTS.utils.manage import ModelManager

LANGUAGE = 'en'
BATCH_SIZE = 4   # T4=4, A100=8
EPOCHS = 200

print('Downloading XTTS v2...')
manager = ModelManager()
model_path, config_path, _ = manager.download_model(
    'tts_models/multilingual/multi-dataset/xtts_v2'
)

config = XttsConfig(
    output_path=str(OUT_DIR),
    model_args=XttsArgs(gpt_use_perceiver_resampler=True),
    audio=XttsAudioConfig(
        sample_rate=22050,
                output_sample_rate=24000,
    ),
    batch_size=BATCH_SIZE,
    eval_batch_size=2,
    num_loader_workers=2,
    eval_split_size=0.1,
    print_step=50,
    plot_step=100,
    log_model_step=1000,
    save_step=5000,
    save_n_checkpoints=2,
    save_checkpoints=True,
    print_eval=False,
    use_phonemes=False,
    languages=[LANGUAGE],
    epochs=EPOCHS,
    datasets=[{
        'formatter': 'ljspeech',
        'meta_file_train': str(DATASET_DIR / 'metadata.csv'),
        'path': str(DATASET_DIR),
        'language': LANGUAGE,
    }],
)

train_samples, eval_samples = load_tts_samples(
    config.datasets, eval_split=True, eval_split_size=config.eval_split_size
)
print(f'Train: {len(train_samples)} | Eval: {len(eval_samples)}')

model = Xtts.init_from_config(config)
model.load_checkpoint(config, checkpoint_dir=model_path, eval=False)

for name, param in model.named_parameters():
    if 'gpt' not in name:
        param.requires_grad = False

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f'Trainable: {trainable:,} / {total:,} params')

trainer = Trainer(
    TrainerArgs(restore_path=None, skip_train_epoch=False,
                start_with_eval=True, grad_clip=0.5),
    config,
    output_path=str(OUT_DIR),
    model=model,
    train_samples=train_samples,
    eval_samples=eval_samples,
)
trainer.fit()
print('Training complete!')
```

```python
# 6 — Test fine-tuned voice
import glob
from TTS.api import TTS
from IPython.display import Audio, display

SPEAKER_WAV = str(sorted((DATASET_DIR / 'wavs').glob('*.wav'))[0])

tts = TTS(model_name='tts_models/multilingual/multi-dataset/xtts_v2')

checkpoints = sorted(glob.glob(str(OUT_DIR / '**/*.pth'), recursive=True))
if checkpoints:
    import torch
    state = torch.load(checkpoints[-1], map_location='cpu')
    if 'model' in state:
        tts.synthesizer.tts_model.load_state_dict(state['model'], strict=False)
    print(f'Loaded: {checkpoints[-1]}')

texts = [
    'Hello, this is Aswini calling from Yexis Electronics.',
    'We have exciting Samsung products for your requirements.',
    'Please let me know if you need any assistance.',
]
for i, text in enumerate(texts):
    out = f'/content/test_{i}.wav'
    tts.tts_to_file(text=text, speaker_wav=SPEAKER_WAV, language='en', file_path=out)
    print(f'[{i}]', text)
    display(Audio(out))
```

```python
# 7 — Save model to Google Drive
import shutil
DRIVE_OUT = '/content/drive/MyDrive/xtts_finetuned'
if os.path.exists(DRIVE_OUT):
    shutil.rmtree(DRIVE_OUT)
shutil.copytree(str(OUT_DIR), DRIVE_OUT)
print(f'Saved to: {DRIVE_OUT}')
print('Files:', os.listdir(DRIVE_OUT))
```
